<h3>Import useful librairies and load csv file created previously</h3>

```python
import pandas as pd
import numpy as np
```

```python
dt = pd.read_csv("dirtytweets.csv")
dt
```

<h3>Let's remove the unnecessary text format</h3>

```python
import emoji
```

```python
dt["tweet"] = dt['tweet'].str.replace("\n", "")
```

```python
dt["tweet"] = dt["tweet"].apply(lambda x : emoji.replace_emoji(x, ""))
```

```python
dt["tweet"] = dt["tweet"].str.replace(r"http\S+|@\S+|#\S+", "", regex=True)
dt
```

<h3>Now le's work on the date and time column</h3>

```python
dt.insert(3, "hour", 0)
dt.insert(4,"year", 0)
dt.insert(7, "year_of_creation", 0)
```

```python
# All the dates are written the same way : 'Day Month Day(digit)' & at the end the year
# So the format is the same : 3 letters for the day + " " + 3 letters for the month + " " + 2 digits for the day
# We want to export the year for the date to the year column created previously
```

```python
dt["hour"] = dt['date'].str[11:16]
dt
```

```python
dt["year"] = dt["date"].str.slice(-4)
```

```python
dt
```

```python
dt.day = dt.date.str[8:10]
dt
```

```python
dt.insert(4,"month",0)
```

```python
dt.month = dt.date.str[4:7]
dt
```

```python
dt.date = dt.date.str.slice(stop=3)
dt
```

```python
dt.year_of_creation = dt.creation.str.slice(-4)
```
